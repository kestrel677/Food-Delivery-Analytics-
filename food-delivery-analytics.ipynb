# D. Answer the 3 Competition Questions

## Traffic Impact:

### Q1 – Traffic Impact: Which road traffic condition has the highest average delivery time?

```python
traffic_summary = (
    df.groupby("Road_traffic_density")["Time_taken (min)"]
    .agg(
        Average_Time="mean",
        Min_Time="min",
        Max_Time="max",
        Total_Orders="count"
    )
    .reset_index()
    .sort_values(by="Average_Time", ascending=False)
)

print("--- Detailed Traffic Impact Analysis ---")
print(traffic_summary.to_string(index=False))
```

**Answer:** 

**Highest Impact Category:** Jam traffic has the highest average delivery time at 31.44 minutes (spanning from a minimum of 10 to a maximum of 54 minutes across 12,414 orders).

**Comparative Insight:** Compared to Low traffic conditions (average of 21.50 minutes across 13,139 orders), heavy traffic jams delay deliveries by nearly 10 minutes per order (~46% increase in travel time).

## Distance Impact:

### Q2 – How does delivery distance affect delivery time? Support your conclusion with data.

```python
df["distance_group"] = pd.cut(
    df["distance_km"],
    bins=[0, 5, 10, 15, 25],
    labels=["0-5 km", "5-10 km", "10-15 km", "15-25 km"]
)

distance_summary = (
    df.groupby("distance_group", observed=False)["Time_taken (min)"]
    .agg(
        Average_Time="mean",
        Min_Time="min",
        Max_Time="max",
        Total_Orders="count"
    )
    .reset_index()
)

print("--- Detailed Distance Impact Analysis ---")
print(distance_summary.to_string(index=False))

correlation = df["distance_km"].corr(df["Time_taken (min)"])
print(f"\nOverall Correlation Coefficient: {correlation:.4f}")
```

**Answer:-**

Statistical Correlation: There is a positive linear correlation ($\approx 0.32$)between distance and delivery time, proving that distance directly slows down deliveries.
* Bracket Breakdown:
* Short trips (0–5 km) average 22.42 minutes across 10,390 orders.
* Medium trips (5–10 km) average 24.64 minutes.
* Long trips (10–15 km and 15–25 km) jump sharply to an average of ~30.09 minutes.* 

##  Combined Conditions:

### Q3 – Which combination of weather condition and traffic density has the highest average delivery time?

```python
combined_summary = (
    df.groupby(["Weather_conditions", "Road_traffic_density"])["Time_taken (min)"]
    .agg(
        Average_Time="mean",
        Min_Time="min",
        Max_Time="max",
        Total_Orders="count"
    )
    .reset_index()
    .sort_values(by="Average_Time", ascending=False)
)

print("--- Detailed Combined Conditions Analysis ---")
print(combined_summary.head(5).to_string(index=False))
```

**Answer:-**

* Highest Impact Combination: The absolute worst operating environment for delivery speed is Foggy weather combined with a Traffic Jam, yielding the highest average delivery time of 36.89 minutes across 2,162 orders.

* Runner-Up Context: Cloudy weather paired with a Traffic Jam is a close second at 36.71 minutes (2,080 orders). Both conditions severely compound delivery friction, extending total delivery duration by over 10 minutes compared to optimal baselines.

# F. Business Insights

##   **Metropolitan Fleet Rebalancing:**

 Metropolitan sectors absorb the overwhelming majority of heavy traffic congestion and gridlock ("Jam"). Implementing dynamic geofencing and surge-pay incentives during peak hours will attract and retain courier capacity where demand peaks, preventing localized fulfillment bottlenecks.

##  **Weather-Adaptive Dynamic SLAs:**

Adverse weather events (fog, storms, and sandstorms) significantly stretch delivery duration distributions and shift the median outward. Transitioning from static estimated times of arrival (ETAs) to an automated, weather-weighted predictive algorithm will safeguard customer satisfaction and protect courier safety metrics.

##  **Smart Multi-Delivery Batching Guardrails:**

Assigning multiple deliveries concurrently introduces compounding logistical delays. Dispatch engines should enforce intelligent batching rules—restricting multi-orders strictly to high-efficiency couriers operating within low-traffic, short-distance corridors.

##  **Proactive Fleet Asset Health Monitoring:** 
Vehicle condition directly impacts failure rates and transit reliability. Integrating telematics-driven maintenance alerts ensures that delivery fleets undergo preventative servicing before mechanical issues disrupt delivery windows during high-volume periods.
